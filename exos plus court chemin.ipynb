```python
#exo 1

def parse_graph(fichier):
    with open(fichier,'r') as f :
        a=f.readlines()
    for i in range(len(a)):
        a[i]=a[i].strip().split(',')
    G={}
    for elt in a:
        G[elt[0]]={}
        G[elt[1]]={}
    for elt in a:
        G[elt[0]][elt[1]]=elt[2]
    return G
```

```python
def step_reachables(graph : dict, s : str , L : list):
    new_v=graph.get(s,{}.keys)
    unseen_v=[]
    for v in new_v:
        if v not in L:
            L.append(v)
            unseen_v.append(v)
    for v in unseen_v:
        step_reachables(graph,v,L)
    return L
```

```python
import math

def voisins(s,dico):           #ça marche
     d=dico.get(s,{}.keys)
     L=[k for k in d]
     for item in dico:
          if s in dico.get(item,{}.keys):
               L.append(item)
     return L

def sommet_plus_proche(liste,dico):
    sommet = liste[0]
    for elt in liste:
         if dico[elt]<dico[sommet]:
              sommet = elt
    return sommet

def longueur_arête(a,b,dico):
     if b not in dico[a].keys:
          return dico
     

def plus_court_chemin(graph,s,z):
    visited=['s']
    if z not in step_reachables(graph,s,[]):
        return 'il n_existe pas de chemin entre ces deux sommets'
    distance={}
    for elt in graph:
            distance[elt[0]]=math.inf
    distance[s]=0
    unseen=voisins(graph,s)
    while 'z' not in visited:
         elt = sommet_plus_proche(unseen,distance)
         visited.append(elt)
         unseen=[]
         for item in voisins(graph,elt):
              if item not in visited:
                   unseen.append(item)
        for item in unseen:
            distance[item] = distance[elt] + longueur_arête(elt,item,graph)
    return distance

```
