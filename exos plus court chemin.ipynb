```python
#exo 1

def parse_graph(fichier):
    with open(fichier,'r') as f :
        a=f.readlines()
    for i in range(len(a)):
        a[i]=a[i].strip().split(',')
    G={}
    for elt in a:
        G[elt[0]]={}
        G[elt[1]]={}
    for elt in a:
        G[elt[0]][elt[1]]=elt[2]
    return G
```

```python
def step_reachables(graph : dict, s : str , L : list):
    new_v=graph.get(s,{}.keys)
    unseen_v=[]
    for v in new_v:
        if v not in L:
            L.append(v)
            unseen_v.append(v)
    for v in unseen_v:
        step_reachables(graph,v,L)
    return L
```
